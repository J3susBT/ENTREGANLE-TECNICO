```python
# =============================================================================
# APLICACIÓN DE MÓDULOS Y PAQUETES EN PYTHON PARA MACHINE LEARNING EN GOOGLE COLAB
# =============================================================================

# 1. INSTALACIÓN DE DEPENDENCIAS Y CONFIGURACIÓN INICIAL
!pip install -q nltk scikit-learn tensorflow torch pandas numpy matplotlib seaborn scipy

import os
import re
import warnings
from collections import Counter

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
from scipy.stats import pearsonr

# Machine Learning
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, classification_report, confusion_matrix

# NLP
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.sentiment import SentimentIntensityAnalyzer

# Deep Learning
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout

warnings.filterwarnings("ignore")
sns.set_style("whitegrid")
plt.rcParams["figure.figsize"] = (12, 6)

# Recursos de NLTK
for resource in ["punkt", "stopwords", "vader_lexicon", "punkt_tab"]:
    try:
        nltk.download(resource, quiet=True)
    except Exception:
        pass

# Crear directorio de datos
os.makedirs("data", exist_ok=True)

print("=" * 70)
print("ENTORNO CONFIGURADO EN GOOGLE COLAB")
print("=" * 70)

# =============================================================================
# 2. CARGA O GENERACIÓN DE DATOS EN GOOGLE COLAB
# =============================================================================

# Si no existen los archivos, creamos muestras de prueba o puedes subirlos
REPORTES_PATH = os.path.join("data", "reportes.txt")
with open(REPORTES_PATH, "w", encoding="utf-8") as f:
    f.write("Los ingresos del sector financiero subieron un 15% este trimestre debido a la estabilidad cambiaria.\n")
    f.write("La inflación generó pérdidas significativas y redujo la rentabilidad operativa anual.\n")
    f.write("Las exportaciones de minerales se mantuvieron estables en comparación al año previo.\n")

print(f"[OK] Archivo {REPORTES_PATH} creado.")

# Para subir tus archivos data.csv y data_peru.csv en Colab directamente:
from google.colab import files
print("\nPor favor sube 'data.csv' y 'data_peru.csv' si aún no están en la carpeta 'data/':")
uploaded = files.upload()

for fn in uploaded.keys():
    os.rename(fn, os.path.join("data", fn))

# =============================================================================
# 3. LECTURA Y PIVOTEO DE PERÚ (FORMATO LARGO A ANCHO)
# =============================================================================

print("\n" + "=" * 70)
print("3. ANÁLISIS DE DATASET DE PERÚ")
print("=" * 70)

DATA_PERU_PATH = os.path.join("data", "data_peru.csv")

if os.path.exists(DATA_PERU_PATH):
    df_peru_raw = pd.read_csv(DATA_PERU_PATH)

    peru_key_indicators = [
        "GDP (current US$)",
        "GDP growth (annual %)",
        "GDP per capita (current US$)",
        "Inflation, consumer prices (annual %)",
        "Exports of goods and services (current US$)",
        "Imports of goods and services (current US$)",
        "Manufacturing, value added (% of GDP)",
        "Services, value added (% of GDP)",
        "Agriculture, forestry, and fishing, value added (% of GDP)"
    ]

    df_peru_filtered = df_peru_raw[
        df_peru_raw["Indicator Name"].isin(peru_key_indicators)
    ][["Year", "Indicator Name", "Value"]]

    df_peru = df_peru_filtered.pivot_table(
        index="Year", columns="Indicator Name", values="Value", aggfunc="first"
    )
    df_peru.columns = [c.strip() for c in df_peru.columns]
    df_peru = df_peru.sort_index()

    print(f"[OK] Dataset Perú procesado: {df_peru.shape[0]} años × {df_peru.shape[1]} indicadores")
    print(df_peru.describe())
else:
    print(f"[AVISO] No se encontró {DATA_PERU_PATH}. Sube el archivo para ejecutar este bloque.")

# =============================================================================
# 4. PROCESAMIENTO DE LENGUAJE NATURAL (NLP) & SENTIMIENTO
# =============================================================================

print("\n" + "=" * 70)
print("4. PROCESAMIENTO DE LENGUAJE NATURAL (NLP)")
print("=" * 70)

with open(REPORTES_PATH, "r", encoding="utf-8") as f:
    sample_reports = [line.strip() for line in f if line.strip()]

def clean_text(text):
    text = text.lower()
    text = re.sub(r"[^a-zA-Z0-9\sáéíóúñ]", "", text)
    return re.sub(r"\s+", " ", text).strip()

cleaned_reports = [clean_text(r) for r in sample_reports]

sia = SentimentIntensityAnalyzer()
print("\n--- Resultados de Sentimiento ---")
for i, report in enumerate(sample_reports):
    scores = sia.polarity_scores(report)
    compound = scores["compound"]
    label = "Positivo" if compound >= 0.05 else ("Negativo" if compound <= -0.05 else "Neutral")
    print(f"Reporte {i+1} [{label}]: {report}")

# =============================================================================
# 5. RED NEURONAL PARA SERIES TEMPORALES (LSTM - TENSORFLOW)
# =============================================================================

print("\n" + "=" * 70)
print("5. MODELO DE RED NEURONAL LSTM (PIB PERÚ)")
print("=" * 70)

if 'df_peru' in locals() and "GDP (current US$)" in df_peru.columns:
    ts_data = df_peru["GDP (current US$)"].dropna().values.reshape(-1, 1)

    if len(ts_data) > 10:
        scaler = StandardScaler()
        ts_scaled = scaler.fit_transform(ts_data)

        lookback = 3
        X_ts, y_ts = [], []
        for i in range(len(ts_scaled) - lookback):
            X_ts.append(ts_scaled[i : i + lookback])
            y_ts.append(ts_scaled[i + lookback])

        X_ts, y_ts = np.array(X_ts), np.array(y_ts)

        split = int(len(X_ts) * 0.8)
        X_train, X_test = X_ts[:split], X_ts[split:]
        y_train, y_test = y_ts[:split], y_ts[split:]

        # Construcción del modelo LSTM
        model = Sequential([
            LSTM(32, input_shape=(lookback, 1), return_sequences=False),
            Dropout(0.2),
            Dense(1)
        ])

        model.compile(optimizer='adam', loss='mse')
        history = model.fit(X_train, y_train, epochs=20, batch_size=4, verbose=0, validation_data=(X_test, y_test))

        print("[OK] Entrenado con éxito el modelo LSTM para predicción de PIB.")
        print(f"Pérdida (MSE) final en Test: {model.evaluate(X_test, y_test, verbose=0):.4f}")
```
